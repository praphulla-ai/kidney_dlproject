```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

load_dotenv()
if not os.getenv("OPENAI_API_KEY"):
    raise ValueError("OPENAI_API_KEY is not set.")

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.2)
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant. Answer clearly and remember conversation context."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])
history = []

def chat(user_input):
    try:
        messages = prompt.format_messages(history=history, input=user_input)
        response = llm.invoke(messages)
        answer = response.content
        history.extend([HumanMessage(content=user_input), AIMessage(content=answer)])
        return answer
    except Exception as exc:
        return f"Error: {exc}"
```

```python
print(chat("My name is Praphulla."))
print(chat("What is my name?"))
print(chat("Give me one Python interview tip."))
```
